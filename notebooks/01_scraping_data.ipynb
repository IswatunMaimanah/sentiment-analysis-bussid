```python
# ==============================
# Scraping Ulasan Bus Simulator Indonesia (com.maleo.bussimulatorid)
# Target: 10.000 ulasan relevan dengan fitur utama
# Periode: 2020–2025
# Bahasa: Indonesia
# ==============================

from google_play_scraper import reviews, Sort
import pandas as pd
from datetime import datetime
import time

# ---------- PARAMETER ----------
APP_ID = 'com.maleo.bussimulatorid'
TARGET_COUNT = 10000  # target jumlah ulasan relevan
LANG = 'id'
COUNTRY = 'id'

# ---------- FITUR UTAMA YANG AKAN DICARI DALAM ULASAN ----------
KEYWORDS = [
    # Desain dan livery
    'livery', 'desain', 'cat bus', 'warna bus',

    # Suasana jalanan dan trayek
    'jalanan', 'trayek', 'rute', 'jalur', 'kota', 'desa',

    # PO Bus dan kendaraan lokal
    'po bus', 'bus asli', 'bus indonesia', 'kendaraan', 'sasis', 'mesin',

    # Kualitas grafis dan tampilan visual
    'grafis', 'visual', 'tampilan', 'detail',

    # Multiplayer dan konvoi
    'konvoi', 'multiplayer', 'online',

    # Suara klakson telolet
    'telolet', 'klakson',

    # Mod kendaraan
    'mod', 'modifikasi',

    # Penyimpanan dan fitur save
    'penyimpanan', 'save', 'otomatis',

    # Iklan dan banner
    'iklan', 'banner', 'tanpa iklan'
]

# ---------- BATAS WAKTU ----------
START_DATE = datetime(2020, 1, 1)
END_DATE = datetime(2025, 12, 31)

# ---------- PROSES SCRAPING ----------
all_relevant_reviews = []
token = None

print("Mulai scraping ulasan relevan...")

def is_relevant(text):
    return any(keyword in text for keyword in KEYWORDS)

while len(all_relevant_reviews) < TARGET_COUNT:
    result, token = reviews(
        APP_ID,
        lang=LANG,
        country=COUNTRY,
        sort=Sort.NEWEST,
        count=200,
        continuation_token=token
    )

    # Konversi ke DataFrame sementara
    df_temp = pd.DataFrame(result)
    df_temp['at'] = pd.to_datetime(df_temp['at'])
    df_temp['content'] = df_temp['content'].astype(str).str.lower()

    # Filter waktu dan keyword
    df_temp = df_temp[
        (df_temp['at'] >= START_DATE) &
        (df_temp['at'] <= END_DATE) &
        (df_temp['content'].apply(is_relevant))
    ]

    all_relevant_reviews.extend(df_temp.to_dict('records'))

    print(f"Total relevan sementara: {len(all_relevant_reviews)} ulasan...")

    if not token:
        print("Token habis (semua ulasan sudah diambil).")
        break

    time.sleep(1)

# ---------- BATASI HINGGA 10.000 SAJA ----------
df_final = pd.DataFrame(all_relevant_reviews).head(TARGET_COUNT)

# ---------- SIMPAN KE FILE ----------
df_final.to_csv('ulasan_bus_simulator_2020_2025_filtered_10000.csv', index=False, encoding='utf-8-sig')

print(f"\n✅ Selesai! Total {len(df_final)} ulasan relevan disimpan ke 'ulasan_bus_simulator_2020_2025_filtered_10000.csv'")
```

```python
# ==============================
# Scraping Ulasan Bus Simulator Indonesia (com.maleo.bussimulatorid)
# Target: 6.000 ulasan relevan (2020–2025)
# Distribusi: ~1.000 ulasan per tahun
# Bahasa: Indonesia
# ==============================

from google_play_scraper import reviews, Sort
import pandas as pd
from datetime import datetime
import time
import math

# ---------- PARAMETER ----------
APP_ID = 'com.maleo.bussimulatorid'
TARGET_TOTAL = 6000           # total target semua tahun
LANG = 'id'
COUNTRY = 'id'

# ---------- FITUR UTAMA ----------
KEYWORDS = [
    'livery', 'desain', 'cat bus', 'warna bus',
    'jalanan', 'trayek', 'rute', 'jalur', 'kota', 'desa',
    'po bus', 'bus asli', 'bus indonesia', 'kendaraan', 'sasis', 'mesin',
    'grafis', 'visual', 'tampilan', 'detail',
    'konvoi', 'multiplayer', 'online',
    'telolet', 'klakson',
    'mod', 'modifikasi',
    'penyimpanan', 'save', 'otomatis',
    'iklan', 'banner', 'tanpa iklan'
]

# ---------- PERIODE ----------
YEARS = [2020, 2021, 2022, 2023, 2024, 2025]
PER_YEAR_TARGET = math.floor(TARGET_TOTAL / len(YEARS))  # ~1667 per tahun

# ---------- FUNGSI FILTER ----------
def is_relevant(text):
    text = text.lower()
    return any(keyword in text for keyword in KEYWORDS)

# ---------- PROSES SCRAPING ----------
all_reviews = []

print("🚀 Mulai scraping ulasan relevan per tahun (2020–2025)...\n")

for year in YEARS:
    print(f"📅 Mengambil ulasan tahun {year}...")
    year_reviews = []
    token = None

    start_date = datetime(year, 1, 1)
    end_date = datetime(year, 12, 31)

    while len(year_reviews) < PER_YEAR_TARGET:
        result, token = reviews(
            APP_ID,
            lang=LANG,
            country=COUNTRY,
            sort=Sort.NEWEST,
            count=200,
            continuation_token=token
        )

        # Konversi ke DataFrame
        df_temp = pd.DataFrame(result)
        df_temp['at'] = pd.to_datetime(df_temp['at'])
        df_temp['content'] = df_temp['content'].astype(str).str.lower()

        # Filter tahun dan relevansi
        df_temp = df_temp[
            (df_temp['at'] >= start_date) &
            (df_temp['at'] <= end_date) &
            (df_temp['content'].apply(is_relevant))
        ]

        # Tambah ke kumpulan tahun
        year_reviews.extend(df_temp.to_dict('records'))

        print(f"   ↳ {len(year_reviews)} ulasan relevan terkumpul ({year})...")

        if not token:
            print(f"⚠️ Token habis untuk tahun {year}, hanya {len(year_reviews)} ulasan yang tersedia.")
            break

        time.sleep(1)  # jeda antar request

    # Batasi sesuai target tahunan
    df_year = pd.DataFrame(year_reviews).head(PER_YEAR_TARGET)
    all_reviews.extend(df_year.to_dict('records'))

    print(f"✅ Tahun {year} selesai ({len(df_year)} ulasan diambil)\n")
    time.sleep(2)

# ---------- GABUNGKAN DAN SIMPAN ----------
df_final = pd.DataFrame(all_reviews).head(TARGET_TOTAL)
df_final.to_csv('ulasan_bus_simulator_2020_2025_filtered_equal_6000.csv', index=False, encoding='utf-8-sig')

print(f"\n🎯 Selesai! Total {len(df_final)} ulasan relevan (≈1.000 per tahun) disimpan ke 'ulasan_bus_simulator_2020_2025_filtered_equal_10000.csv'")
```

```python
# ==============================
# Scraping Ulasan Bus Simulator Indonesia (com.maleo.bussimulatorid)
# Periode: 2021–2025 (semua ulasan relevan)
# Bahasa: Indonesia
# ==============================

from google_play_scraper import reviews, Sort
import pandas as pd
from datetime import datetime
import time

# ---------- PARAMETER ----------
APP_ID = 'com.maleo.bussimulatorid'
LANG = 'id'
COUNTRY = 'id'

# ---------- FITUR UTAMA ----------
KEYWORDS = [
    'livery', 'desain', 'cat bus', 'warna bus',
    'jalanan', 'trayek', 'rute', 'jalur', 'kota', 'desa',
    'po bus', 'bus asli', 'bus indonesia', 'kendaraan', 'sasis', 'mesin',
    'grafis', 'visual', 'tampilan', 'detail',
    'konvoi', 'multiplayer', 'online',
    'telolet', 'klakson',
    'mod', 'modifikasi',
    'penyimpanan', 'save', 'otomatis',
    'iklan', 'banner', 'tanpa iklan'
]

# ---------- PERIODE ----------
YEARS = [2021, 2022, 2023, 2024, 2025]

# ---------- FUNGSI FILTER ----------
def is_relevant(text):
    text = text.lower()
    return any(keyword in text for keyword in KEYWORDS)

# ---------- PROSES SCRAPING ----------
all_reviews = []

print("🚀 Mulai scraping semua ulasan relevan (2021–2025)...\n")

token = None
finished = False

while not finished:
    result, token = reviews(
        APP_ID,
        lang=LANG,
        country=COUNTRY,
        sort=Sort.NEWEST,
        count=200,
        continuation_token=token
    )

    if not result:
        break

    df_temp = pd.DataFrame(result)
    df_temp['at'] = pd.to_datetime(df_temp['at'])
    df_temp['content'] = df_temp['content'].astype(str).str.lower()

    # Filter tahun 2021–2025
    df_temp = df_temp[
        (df_temp['at'].dt.year >= 2021) &
        (df_temp['at'].dt.year <= 2025)
    ]

    # Filter relevansi keyword
    df_temp = df_temp[df_temp['content'].apply(is_relevant)]

    # Simpan
    all_reviews.extend(df_temp.to_dict('records'))

    print(f"📌 Total sementara: {len(all_reviews)} ulasan relevan...")

    if not token:
        finished = True

    time.sleep(1)

print("\n⏳ Scraping selesai, memproses data...")

# ---------- SIMPAN ----------
df_final = pd.DataFrame(all_reviews)
df_final.to_csv('ulasan_bus_simulator_2021_2025_filtered_full.csv', index=False, encoding='utf-8-sig')

print(f"\n🎯 Selesai! Total {len(df_final)} ulasan relevan telah disimpan ke:")
print("➡️  'ulasan_bus_simulator_2021_2025_filtered_full.csv'")
```

```python
# Pastikan tanggal dibaca dengan benar
df['at'] = pd.to_datetime(df['at'])

# ==============================
# BUAT KOLOM YEAR-MONTH (YYYY-MM)
# ==============================
df['year_month'] = df['at'].dt.to_period('M').astype(str)

# ==============================
# HITUNG JUMLAH ULASAN PER BULAN
# ==============================
monthly_count = df.groupby('year_month').size().reset_index(name='jumlah_ulasan')

# ==============================
# HITUNG 20% DARI JUMLAH ULASAN (dibulatkan)
# ==============================
monthly_count['persen_20'] = (monthly_count['jumlah_ulasan'] * 0.20).apply(math.ceil)

# ==============================
# SIMPAN HASIL LENGKAP
# ==============================
monthly_count.to_csv('jumlah_ulasan_perbulan_20persen.csv',
                     index=False, encoding='utf-8-sig')

print("📊 Rekap jumlah ulasan + 20% per bulan disimpan ke:")
print("➡️ jumlah_ulasan_perbulan_20persen.csv")

# Tampilkan 60 baris
monthly_count.head(60)
```

```python
# ==============================
# AMBIL DATA SESUAI JUMLAH persen_20
# ==============================

# DataFrame untuk menyimpan hasil akhir
sampled_data = pd.DataFrame()

for index, row in monthly_count.iterrows():
    ym = row['year_month']
    n_sample = row['persen_20']

    # Filter data sesuai tahun-bulan
    df_filtered = df[df['year_month'] == ym]

    # Jika sample lebih besar dari jumlah data (sangat jarang), ambil semua
    if n_sample > len(df_filtered):
        n_sample = len(df_filtered)

    # Ambil sample acak
    df_sample = df_filtered.sample(n=n_sample, random_state=42)

    # Tambahkan ke dataframe hasil
    sampled_data = pd.concat([sampled_data, df_sample], ignore_index=True)

# Simpan ke file CSV
sampled_data.to_csv("sample_ulasan_20persen_perbulan.csv", index=False, encoding="utf-8-sig")

print("✓ Data sample 20% per bulan berhasil dibuat.")
print("➡️ File: sample_ulasan_20persen_perbulan.csv")

# Tampilkan 10 baris pertama hasil
sampled_data.head(10)
```
